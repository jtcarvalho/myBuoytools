
# Estrutura de Diretórios e Comandos para Gerar um Pacote com o Poetry

## Estrutura do Repositório

A estrutura recomendada para seu repositório com o Poetry seria a seguinte:

```
my_project/
│
├── my_package/               # Diretório do código-fonte do pacote
│   ├── __init__.py           # Arquivo que define o pacote
│   └── <outros_scripts.py>   # Outros scripts ou módulos
│
├── tests/                    # Diretório dos testes
│   ├── test_my_package.py    # Testes do pacote
│
├── pyproject.toml            # Arquivo de configuração do Poetry
├── README.md                 # Documentação do projeto
└── MANIFEST.in               # Arquivo para incluir arquivos no pacote (opcional)
```

## Passos para Gerar o Pacote com o Poetry

### 1. Inicializar o Projeto com Poetry

Primeiro, inicialize o projeto usando o Poetry com o comando:

```bash
poetry init
```

Isso cria o arquivo `pyproject.toml`, onde você irá definir as dependências, nome do pacote, versão, descrição e outros detalhes. Durante o processo, o Poetry vai perguntar se você deseja adicionar dependências imediatamente.

### 2. Instalar as Dependências

#### 2.1
Após o `poetry init`, o próximo passo é instalar as dependências necessárias. O comando para isso é:

```bash
poetry install
```
Isso criará o ambiente virtual e instalará as dependências listadas no arquivo `pyproject.toml`.

#### 2.2

Outra opcao, que eu prefiro, é instalar o ambiente virtual desta maneira:

```bash
 python -m venv py312sushi
```
Depois ativar o ambiente:

```bash
 source py312sushi/bin/activate
```

### 3. Adicionar Dependências

Se você precisar adicionar dependências adicionais ao seu projeto, pode usar o comando:

```bash
poetry add <nome_do_pacote>
```

Todas essas dependencia serao incluidas no ambiente mesmo sendo gerado pelo venv e sera
adiciona ao arquivo pyproject.toml automaticamente.


### `4. Ativar o Ambiente Virtual`


`Se nao ativou com o venv, para ativar o ambiente virtual criado pelo Poetry, execute:`

```bash
poetry shell
``` 



### 5. Criar o Pacote

Após configurar o projeto e instalar as dependências, o próximo passo é criar o pacote. Para gerar o pacote e as distribuições, use o comando:

```bash
poetry build
```

Isso criará os arquivos `.tar.gz` e `.whl` dentro da pasta `dist/`.


### 6. Testar localmente

Para testar o pacote localmente antes de publicar, executar:

```bash
pip install .
```

### 7. Atualizar libs e pacote

Caso seja necessario atualizar ou adicionar uma nova lib, deve-se ativar o ambiente virtual e:

```bash
pip add <nome_do_pacote>
```

```bash
pip update
```

```bash
pip check
```

Se necessario reinstalar tudo:
```bash
pip install 
```
Instalamos localmente para testar e depois seguimos para o passo 8 novamente 

### 8. Publicar o Pacote no PyPI

Para publicar o pacote no PyPI, use o comando:

```bash
poetry publish --build
```

Isso vai gerar o pacote e publicar diretamente no PyPI. Você precisará configurar suas credenciais do PyPI, seja através de login e senha ou um token.

### 9. Instalar

```bash
pip install seasushi
```

depois de instalado testar um teste padrão:

### Gerenciamento de Versões com Poetry

 📌 O que significa `0.1.0`?
  O número da versão segue a convenção Semantic Versioning (SemVer):

 MAJOR.MINOR.PATCH

 - `0` → Versão principal (MAJOR): Mudança **incompatível** na API.
 - `1` → Versão secundária (MINOR): Novas funcionalidades **compatíveis**.
 - `0` → Correção de bugs (PATCH): Pequenas correções **sem quebrar compatibilidade**.

 Exemplo de versões:
 - `1.0.0` → Primeira versão estável
 - `1.1.0` → Novas funcionalidades compatíveis
 - `1.1.1` → Apenas correções de bugs


✅ 1. Atualizar a versão do pacote automaticamente
 Você pode atualizar a versão do pacote usando Poetry:

 🔹 Aumentar versão de **patch** (correção de bug):
```bash
poetry version patch
```
Exemplo: 0.1.0 → 0.1.1

🔹 Aumentar versão de **minor** (novos recursos compatíveis):

```bash
poetry version minor
```

Exemplo: 0.1.0 → 0.2.0

🔹 Aumentar versão de **major** (mudanças incompatíveis):

```bash
poetry version major
```

Exemplo: 0.1.0 → 1.0.0


 ✅ 2. Adicionar tags Git para versionamento (opcional)
  Se estiver usando Git, é uma boa prática versionar seu código:

```bash
git commit -am "Lançamento da versão 0.1.0: Adicionada nova funcionalidade"
```

```bash
git tag v0.1.0
```

```bash
git push origin v0.1.0 ou 
git push --tags
```

 ✅ 3. Publicar no PyPI
Após definir a versão correta, publique seu pacote:

```bash
poetry publish --build
```
